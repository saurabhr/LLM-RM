```python
import rmllm
import pandas as pd
```

```python
fig_path = rmllm.config.FIGURES_DIR/"svgs"
fig_path.mkdir(parents=True, exist_ok=True)

data_dir = rmllm.config.PROCESSED_DATA_DIR

exp_1 = pd.read_csv(data_dir / "rmai_exp_1.csv")
exp_2 = pd.read_csv(data_dir / "rmai_exp100_2.csv")
exp_2_r = pd.read_csv(data_dir / "rmai_exp100_2_r.csv")

```

```python

model_map_1 = {
    "gemma3:12b":"Gemma3:12b",
    "gemma3:12b-it-qat":"Gemma3:12b-QAT",
    "gemma3:27b":"Gemma3:27b",
    "gemma3:27b-it-qat":"Gemma3:27b-QAT",
    "llama4:scout":"Llama4:16x17b",
    "llama3.3:70b":"Llama3.3:70b"
}
model_map_2 = {
    "gemma3_12b":"Gemma3:12b",
    "gemma3_12b-it-qat":"Gemma3:12b-QAT",
    "gemma3_27b":"Gemma3:27b",
    "gemma3_27b-it-qat":"Gemma3:27b-QAT",
    "llama4_scout":"Llama4:16x17b",
    "llama3.3_70b":"Llama3.3:70b"
}
exp_1["model"] = exp_1["model"].map(model_map_1)
exp_2["model"] = exp_2["model"].map(model_map_2)


exp_1["accuracy"] = exp_1["accuracy"].astype(int)
exp_2["accuracy"] = exp_2["accuracy"].astype(int)

exp_1["memory"] = exp_1["memory"].map({'SingleTurn':"SingleTurn", 'Convo':"TrialChain"})

exp_1["trial_compliance"] = exp_1["word2acc"].astype(int)
exp_1["reading_hallucination"] = 1 - exp_1["word2acc"].astype(int)

exp_2["trial_compliance"] = exp_2["word2acc"].astype(int)

exp_2["reading_hallucination"] = 1 - exp_2["word2acc"].astype(int)

exp_2["test_source"] = exp_2["source_test"]
exp_2["gen_source"] = exp_2["source_ort"]

exp_1.rename(columns={"Rating":"rating"},inplace=True)
exp_1.rename(columns={"Confidence":"confidence"},inplace=True)

exp_2.rename(columns={"Confidence_test":"confidence"},inplace=True)
exp_2.rename(columns={"Rating_ort":"rating"},inplace=True)
```

```python
exp_1_grp = ["memory", "model", "source"]
exp_1["total_trials"] = exp_1.groupby(exp_1_grp)["trial_compliance"].transform("count")
exp_2_grp = ["trace","model","setsize","fb_exp","test_source"]
exp_2["total_trials"] = exp_2.groupby(exp_2_grp)["trial_compliance"].transform("count")
```

# Prepare Source Accuracy DataFrames for Analysis

```python
# Experiment 1
exp1_src_accuracy = exp_1.groupby(exp_1_grp)[["total_trials","rating","reading_hallucination","accuracy","confidence"]].mean().reset_index()
```

```python
# Experiment 2
exp2_src_accuracy = exp_2.pivot_table(index=exp_2_grp,values = ["total_trials","rating","reading_hallucination","accuracy","confidence"],aggfunc="mean").reset_index()
```

```python
# Experiment 1
exp1_src_accuracy = exp_1.groupby(exp_1_grp).agg({
    'total_trials': 'first',  # Or 'mean' since it's constant per group
    'rating': 'mean',
    'reading_hallucination': ['mean', 'sum'],
    'accuracy': ['mean', 'sum'],
    'confidence': 'mean'
}).reset_index()

# Flatten multi-level columns
exp1_src_accuracy.columns = ['_'.join(col).strip() if col[1] else col[0] for col in exp1_src_accuracy.columns.values]

# Experiment 2
aggfunc = {
    'total_trials': 'first',  # Or 'mean' since it's constant per group
    'rating': 'mean',
    'reading_hallucination': ['mean', 'sum'],
    'accuracy': ['mean', 'sum'],
    'confidence': 'mean'
}
exp2_src_accuracy = exp_2.pivot_table(index=exp_2_grp, values=['total_trials', 'rating', 'reading_hallucination', 'accuracy', 'confidence'], aggfunc=aggfunc).reset_index()

# Flatten multi-level columns
exp2_src_accuracy.columns = ['_'.join(col).strip() if col[1] else col[0] for col in exp2_src_accuracy.columns.values]
```

# Prepare cumulative trial accuracy data for experiment 2

```python
btw_groups2 = ["setsize","fb_exp","model","trace"]
exp_2["trial_count"] = exp_2.groupby(btw_groups2).cumcount()+1
exp_2["cum_acc_sum"] = exp_2.groupby(btw_groups2)["accuracy"].cumsum()
exp_2["cum_acc"] = exp_2["cum_acc_sum"]/exp_2["trial_count"]
exp_2_20 = exp_2.loc[exp_2["setsize"]==20]
exp_2_40 = exp_2.loc[exp_2["setsize"]==40]
```

# SAVE DATA

```python
exp_1.to_csv(rmllm.config.PROCESSED_DATA_DIR/"exp1_trial_data.csv")
exp_2.to_csv(rmllm.config.PROCESSED_DATA_DIR/"exp2_trial_data.csv")

exp1_src_accuracy.to_csv(rmllm.config.PROCESSED_DATA_DIR/"exp1_group_level.csv")
exp2_src_accuracy.to_csv(rmllm.config.PROCESSED_DATA_DIR/"exp2_group_level.csv")
```
